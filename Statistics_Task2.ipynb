10) Step 1: Total outcomes

    A coin is tossed three times.

    Total outcomes = 2x2x2 = 8

    Possible outcomes:

    {HHH, HHT, HTH, HTT, THH, THT, TTH, TTT}

Step 2: Favourable outcomes

    Exactly two heads:

    {HHT, HTH, THH}

    Favourable outcomes = 3

    Theoretical probability:

    = 3/8 = 0.375

Step 3: Python Simulation

```python
import random
n = 100000
count = 0
for i in range(n):
  tossess = [random.choice(['H', 'T']) for _ in range(3)]

  if tossess.count('H') == 2:
    count += 1

probability = count / n

print("estimated Probability:", probability)
print("Theoretical Probability:", 3/8)
```

Step 4: Which condition changes?

For exactly three heads:

tossess.count('H') == 3

For exactly two heads:

tossess.count('H') == 2

Only the condition changes from 3 heads to 2 heads.

The estimated probability will be approximately 0.375.

7) The researcher is facing Selection Bias because only students from the      highest-performing batch were selected.

  The sample does not represent the entire student population.

  Improved Selection Process:

  Select students randomly from all batches.
  Include students with different performance levels.
  Use stratified random sampling to ensure proper representation.
  Conclusion: A representative random sample from all batches will provide more reliable results.

```python
#6
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
sample_sizes = [2, 10, 50, 200]
repetitions = 10000

for n in sample_sizes:
  sample_means = []
  for i in range(repetitions):
    data = np.random.exponential(scale=10, size=n)
    sample_means.append(np.mean(data))

simulated_se = np.std(sample_means)
theoretical_se = 10 / np.sqrt(n)

print("Sample Size:", n)
print("Mean:", np.mean(sample_means))
print("Simulated SE:", simulated_se)
print("Theoretical SE:", theoretical_se)

plt.hist(sample_means, bins=30)
plt.title(f"Sampling Distribution (n={n})")
plt.xlabel("Sample Means")
plt.ylabel("Frequency")
plt.show()
```
